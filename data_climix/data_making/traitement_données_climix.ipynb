```python
import pandas as pd
from pathlib import Path

# Configuration
base_path = Path(r"C:\Users\anais\OneDrive - insa-toulouse.fr\Bureau\INSA\4A MA\RI\perso\data_for_climix")
years = range(2018, 2024)
# Ajout de wind_offshore à la liste
energy_types = ["solar", "wind_onshore", "wind_offshore"]

# On récupère les noms des dossiers de régions
regions = [d.name for d in base_path.iterdir() if d.is_dir()]

for year in years:
    for energy in energy_types:
        all_region_data = []
        
        for region in regions:
            file_path = base_path / region / str(year) / f"{energy}.csv"
            
            if file_path.exists():
                try:
                    # Lecture auto du séparateur (virgule ou point-virgule)
                    df = pd.read_csv(file_path, sep=None, engine='python')
                    
                    # Nettoyage et normalisation des colonnes
                    df.columns = df.columns.str.strip()
                    df = df.rename(columns={'heure': 'timestamp'})
                    
                    if 'timestamp' in df.columns:
                        df['timestamp'] = pd.to_datetime(df['timestamp'])
                        all_region_data.append(df)
                except Exception as e:
                    print(f"❌ Erreur lors de la lecture de {file_path}: {e}")
            
        # On ne crée le fichier que si on a trouvé au moins une région
        if all_region_data:
            france_df = pd.concat(all_region_data)
            
            # Somme des données par timestamp
            # On utilise 'facteur_charge' (à adapter si le nom varie aussi)
            france_total = france_df.groupby('timestamp')['facteur_charge'].sum().reset_index()
            
            output_filename = f"{energy}_{year}.csv"
            france_total.to_csv(output_filename, index=False)
            print(f"✅ Fichier créé : {output_filename} (Données de {len(all_region_data)} régions)")
        else:
            # Utile pour wind_offshore qui n'existe pas partout
            print(f"ℹ️ Aucune donnée trouvée pour {energy} en {year}")

print("\nTraitement de l'ensemble des énergies terminé.")
```

```python
import pandas as pd
import os

def fusionner_donnees_eoliennes(annees):
    for annee in annees:
        fichier_onshore = f"wind_onshore_{annee}.csv"
        fichier_offshore = f"wind_offshore_{annee}.csv"
        fichier_sortie = f"wind_{annee}.csv"
        
        # Vérification de l'existence des deux fichiers
        if os.path.exists(fichier_onshore) and os.path.exists(fichier_offshore):
            print(f"Traitement de l'année {annee}...")
            
            # Lecture des fichiers CSV
            df_on = pd.read_csv(fichier_onshore)
            df_off = pd.read_csv(fichier_offshore)
            
            # Fusion des DataFrames sur la colonne 'timestamp'
            # On utilise les suffixes pour différencier les colonnes temporairement
            df_fusion = pd.merge(df_on, df_off, on='timestamp', suffixes=('_on', '_off'))
            
            # Calcul de la somme du facteur de charge
            df_fusion['facteur_charge'] = df_fusion['facteur_charge_on'] + df_fusion['facteur_charge_off']
            
            # Sélection des colonnes finales
            df_final = df_fusion[['timestamp', 'facteur_charge']]
            
            # Sauvegarde dans le nouveau fichier
            df_final.to_csv(fichier_sortie, index=False)
            print(f"Fichier créé : {fichier_sortie}")
        else:
            print(f"Avertissement : Fichiers manquants pour l'année {annee}")

# Exécution pour la période 2018-2023
annees_a_traiter = range(2018, 2024)
fusionner_donnees_eoliennes(annees_a_traiter)
```
