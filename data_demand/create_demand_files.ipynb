```python
import pandas as pd
import os

INPUT = 'demand.xlsx'
OUTPUT_DIR = '.'

df = pd.read_excel(INPUT, header=None)
split = df[0].str.split(',', expand=True)
split.columns = ['country', 'hour_id', 'demand']

for country, group in split.groupby('country'):
    group = group.reset_index(drop=True)
    out = pd.DataFrame({
        'hour': range(len(group)),
        'demand': group['demand'].values
    })
    filename = os.path.join(OUTPUT_DIR, f"demand_{country}.csv")
    out.to_csv(filename, index=False, header=False)
    print(f"Created {filename} ({len(out)} rows)")
```
