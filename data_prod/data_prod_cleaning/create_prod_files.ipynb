```python
import pandas as pd
import os

def organize_production_data(input_file):
    # Chargement des données (on suppose qu'il n'y a pas d'en-tête dans le fichier source)
    df = pd.read_csv(input_file, header=None, names=['Country', 'Technology', 'Timestamp', 'Value'])
    
    countries = df['Country'].unique()
    
    # Chemin vers le dossier parent (un niveau au-dessus du script)
    parent_dir = "../countries_data"

    for country in countries:
        # Création du dossier du pays dans le dossier parent
        country_dir = os.path.join(parent_dir, country)
        os.makedirs(country_dir, exist_ok=True)
        
        # Filtrage des données pour le pays actuel
        df_country = df[df['Country'] == country]
        technologies = df_country['Technology'].unique()
        
        # 1. Sauvegarde des fichiers par technologie individuelle
        for tech in technologies:
            # On récupère seulement la colonne 'Value'
            df_tech = df_country[df_country['Technology'] == tech][['Value']].copy()
            
            # On réinitialise l'index pour qu'il commence à 0
            df_tech = df_tech.reset_index(drop=True)
            
            # Transformation du nom : pv -> solar
            tech_label = "solar" if tech == "pv" else tech
            file_name = f"{tech_label}_{country}.csv"
            
            output_path = os.path.join(country_dir, file_name)
            
            # Sauvegarde : index=True pour avoir le numéro (0, 1, 2...), header=False pour ne pas avoir de titre de colonne
            df_tech.to_csv(output_path, header=False, index=True)
            print(f"Créé : {output_path}")
        
        # 2. Création du fichier regroupant on-shore et off-shore (Wind)
        df_combined = df_country[df_country['Technology'].isin(['onshore', 'offshore'])][['Value']].copy()
        if not df_combined.empty:
            df_combined = df_combined.reset_index(drop=True)
            combined_path = os.path.join(country_dir, f"wind_{country}.csv")
            df_combined.to_csv(combined_path, header=False, index=True)
            print(f"Créé : {combined_path} (combiné)")

# Exécution du script
organize_production_data('vre_profiles.csv')
```
